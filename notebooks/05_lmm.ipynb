```python
from huggingface_hub import InferenceClient
import os

HF_TOKEN = os.environ.get("HF_TOKEN")

client = InferenceClient(
    model="meta-llama/Llama-3.2-3B-Instruct",
    token=HF_TOKEN
)
def generate_report(prediction_label, confidence_score):
    system_prompt = """You are a clinical radiology assistant specialized in kidney stone detection.
    Generate a concise professional clinical report based ONLY on the information provided (Include confidence score).
    DO NOT invent, assume, or add any details not explicitly given (no size or location).
    Structure the report with:
    - Findings / Impression
    - Recommendation
    Keep it under 200 words."""

    user_prompt = f"""
    CT Scan Analysis Result:
    - Detection: {prediction_label}
    - Model Confidence: {confidence_score}%

    Generate a clinical report using ONLY the above information.
    Do not add any details that were not provided.
    """

    response = client.chat_completion(
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user",   "content": user_prompt}
        ],
        max_tokens=300,
        temperature=0.1
    )

    return response.choices[0].message.content
```

```python
# Test
report = generate_report("Stone", 94.3)
print(report)
```
